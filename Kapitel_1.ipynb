<font color='orange'> 7. Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när respektive datakategori kan uppstå. </font>

Svar:

Tvärsnittsdata innebär att data samlas in på flera olika individer eller enheter (exempelvis människor, länder eller företag) vid en och samma tidpunkt.

Tidsseriedata innebär att data samlas in för en enskild individ eller enhet över flera olika tidpunkter. tex Registrering av dagliga aktiekurser för ett enskilt företag över ett år, eller mätning av den dagliga temperaturen eller försäljningsvolymen för en produkt över tid

Paneldata innebär att data följer flera olika individer eller enheter över flera tidpunkter tex följa samma grupp av 100 personer årligen under en tioårsperiod för att mäta deras löneutveckling, eller att studera ekonomiska nyckeltal (som BNP och inflation) för flera olika länder över flera påvarandra följande år

<font color='orange'> 9. Generellt sett gäller det att högre är bättre i scikit-learn scoring, därför
används exempelvis scoring='neg_mean_squared_error'. Förklara logiken bakom detta, det vill säga att vi använder “negative” mean squared error. </font>

svar: Eftersom MSE är ett felmått där ett lägre värde egentligen är bättre, multiplicerar scikit-learn resultatet med -1. På så sätt blir ett mindre fel (t.ex. 10 istället för 50) ett högre negativt tal (-10 istället för -50), vilket gör att systemets maximering automatiskt väljer den modell som gör minst fel.

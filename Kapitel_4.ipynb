<font color='orange'> 2. Förklara hur *OvR-* och *OvO*-algoritmerna fungerar. </font>

Svar:

OvR (One-vs-Rest) -> är att för varje klass i ett multiklass problem så tränas modellen för att kunna skilja på om det datapunkten tillhör klassen eller inte vilket då är någon av resterande men det kollar inte vilken specifikt av resterande. exempelis är bollen du tar upp i ett bollhav röd? så kan det finnas flera färger till att gissa.

OvO (One-vs-One) -> anväder istället en modell för varje par av klasser. varje modell jämför klasspar och predikterar en av de två.

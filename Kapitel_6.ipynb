## Faktafrågor

<font color='orange'> 1. Vad är klustring för något? Ge några exempel på tillämpningsområden. </font>

Svar:
Klustring är en oövervakad inlärningsmetod där man grupperar datapunkter utifrån hur lika de är varandra, utan några facit/etiketter på vilken grupp de "egentligen" tillhör. Målet är att punkter inom samma kluster ska vara så lika varandra som möjligt, och punkter i olika kluster så olika som möjligt.

Exempel på tillämpningsområden: kundsegmentering, anomalidetektion, bildsegmentering/komprimering, dokumentklustring (gruppera liknande texter) samt som förbearbetningssteg inför andra ML-modeller.

<font color='orange'> 2. Förklara översiktligt hur K-means fungerar. Använd figur 6.3 (sidan 238)
och figur 6.4 (sidan 239) i din förklaring. </font>

Svar:
K-means fungerar i korthet så här: man väljer ett antal kluster k och k slumpmässiga startcentroider. Varje datapunkt tilldelas sedan till den centroid den ligger närmast (baserat på avstånd, ofta euklidiskt). Därefter flyttas varje centroid till medelvärdet av de punkter som tilldelats den. Tilldelning och uppdatering upprepas tills centroiderna slutar flytta sig (algoritmen konvergerar).

Figur 6.3 visar simulerad data med kända, sanna kluster i fyra olika scenarion: väl separerade sfäriska kluster, icke-sfäriska (avlånga) kluster, kluster med olika varians/spridning, samt kluster av olika storlek. Figur 6.4 visar hur K-means faktiskt klustrar samma data. Eftersom K-means bygger på avstånd till en centroid antar den implicit att kluster är sfäriska och ungefär lika stora/lika spridda, det syns tydligt att K-means presterar dåligt när dessa antaganden inte stämmer (icke-sfäriska kluster, olika varians, olika storlek) samt när man valt fel antal kluster.

<font color='orange'> 3. Hur kan man välja vilket antal kluster som ska användas för en K-means
modell? Använd inertia, silhouette score och silhouette diagram i ditt svar. </font>

Svar:
Armbågsmetoden: plotta inertia (summan av kvadrerade avstånd till närmaste centroid) mot k, och leta efter "armbågen" där kurvan planar ut, ibland otydlig.

Silhouette score: mäter hur nära en punkt ligger sitt eget kluster jämfört med näst närmaste kluster (-1 till 1, högre bättre). Man väljer det k som maximerar medelvärdet.

Silhouette-diagrammet visar värdet per punkt och kluster istället för bara ett medelvärde, vilket avslöjar dåligt separerade eller ojämnt stora kluster som medelvärdet annars kan dölja.

Koduppgifter </font>

Svar:
I ett silhouette-diagram är varje "kniv" ett kluster, bredden på kniven motsvarar hur många punkter klustret innehåller, och den streckade linjen är medelvärdet för silhouette score över alla punkter. Man vill helst att alla knivar ska sträcka sig förbi den streckade linjen (dvs klustret presterar minst lika bra som snittet) och att knivarna är ungefär lika breda (jämnstora kluster), vilket K-means generellt mår bra av.

I figur 6.13 ger k=6 sämst resultat, lägre silhouette score och inte alla knivar når förbi linjen. Vid k=3 och k=4 blir ett av klustren (index 1) väldigt stort jämfört med de andra. k=5 har visserligen en något lägre silhouette score än k=3/k=4, men väljs ändå eftersom klustren där är mer jämnstora. Slutsatsen är att valet inte är en exakt vetenskap, k=3 eller k=4 hade också kunnat vara rimliga val, utan man behöver väga samman score och klusterstorlek med eget omdöme.

c) Klustren kan användas för marknadssegmentering, tex ett fastighetsbolag eller en bank kan rikta olika erbjudanden, prissättning eller marknadsföring till olika kluster baserat på inkomstnivå och geografiskt läge, istället för att behandla alla kunder likadant.

d) Vi undersöker hur många kluster som vore rimligt genom att titta på inertia och silhouette score för olika k (samma tre features som i kodexemplet).

````python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

df = pd.read_csv('dataset/housing.csv')
X = df.loc[:, ['median_income', 'latitude', 'longitude']]

for k in range(2, 9):
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X)
    sil = silhouette_score(X, km.labels_)
    print(f'k={k}: inertia={km.inertia_:.0f}, silhouette={sil:.4f}')
````

Silhouette score är faktiskt högst för k=2 och avtar sedan (med en liten uppgång vid k=6), vilket beror på att latitude/longitude dominerar avståndsberäkningen. Rent statistiskt hade alltså färre kluster kunnat motiveras, men för marknadsundersökningens syfte (att hitta hanterbara, meningsfulla kundsegment) kan ändå fler kluster som k=6 vara ett praktiskt val, återigen ingen exakt vetenskap.

````python
import kagglehub

path = kagglehub.dataset_download('thebumpkin/1980s-classic-hits-with-spotify-data')
print(path)
````

````python
import pandas as pd

df = pd.read_csv(path + '/1980sClassics.csv')
print(df.shape)
df.head()
````

````python
print('Antal dubbletter:', df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
print('Antal rader efter borttagning:', df.shape[0])
````

````python
from sklearn.preprocessing import StandardScaler

features = [
    'Danceability', 'Energy', 'Loudness', 'Speechiness', 'Acousticness',
    'Instrumentalness', 'Liveness', 'Valence', 'Tempo',
]
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
````

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 9):
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X_scaled)
    sil = silhouette_score(X_scaled, km.labels_)
    print(f'k={k}: inertia={km.inertia_:.0f}, silhouette={sil:.4f}')
````

````python
km = KMeans(n_clusters=4, random_state=42, n_init=10).fit(X_scaled)
df['Cluster'] = km.labels_
df['Cluster'].value_counts().sort_index()
````

````python
df.groupby('Cluster')[features].mean().round(2).T
````

````python
for c in sorted(df['Cluster'].unique()):
    top = df[df['Cluster'] == c].sort_values('Popularity', ascending=False).head(3)
    print(f'--- Kluster {c} ---')
    print(top[['Track', 'Artist', 'Popularity']].to_string(index=False))
    print()
````

<font color='orange'>2. Förklara utvärderingsmåtten *RMSE*, *MSE* och *MAE*.</font>

Svar:
- RMSE: Root mean squared error, används mest i samband med att regressionsmodeller utvärderas. Hårdast mod extrema värden då man drar roten ur på talen.
- MSE: Mean squared error, används om vi bara är ute efter rangordningen och inte talet. detta då det är mer snabbare pga att kvadratroten inte används
- MAE: Mean absolute error, tar fram ett absolutbelopp. Detta är då ett tal som räknar totala missen. exempel: |2| = 2 och |-2| = 2.
